```python
#too much null
df.drop(columns='tanggal_menjadi_anggota', inplace=True)
```

```python
#melihat isi dari kolom pendidikan & status pernikahan untuk dilakukan perbaikan
pendidikan_unique = df['pendidikan'].unique()
status_pernikahan_unique = df['status_pernikahan'].unique()

print(f"Kolom pendidikan: {pendidikan_unique}")
print(f"Kolom status pernikahan: {status_pernikahan_unique}")
```

```python
#mengubah isi value dari kedua kolom menjadi NaN
df['pendidikan'] = df['pendidikan'].replace('5', np.nan)
df['status_pernikahan'] = df['status_pernikahan'].replace('5', np.nan)
```

```python
#dilakukan label encoding
label_encoding = {
    'SMP': 0,
    'SMA': 1,
    'Sarjana': 2,
    'Magister': 3,
    'Dokter': 4,
}

df['pendidikan'] = df['pendidikan'].map(label_encoding)
```

```python
label_mapping = {
    'Sendiri': 1,
    'Rencana Menikah': 2,
    'Cerai': 3,
    'Menikah': 4,
    'Cerai Mati': 5,
}

df['status_pernikahan'] = df['status_pernikahan'].map(label_mapping)
```

```python
#imputasi dengan median (bisa diubah menggunakan modus, median ataupun mean)
df = df.fillna(df.median(numeric_only=True))
```

```python
#setelah di imputasi, lihat lagi apakah masih ada kolom kosong/tidak
df.isnull().sum()
```

```python
#lihat statistik
df.describe().T
```

```python
df.plot(kind='box', subplots=True, layout=(-1,3), figsize=(12,8))
plt.tight_layout()
plt.show()
```

```python
#handling outliers
cols = ['pendapatan', 'pembelian_toko']
mask = True

for col in cols:
    Q1 = df[col].quantile(0.25) #25%
    Q3 = df[col].quantile(0.75) #75%
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    mask = mask & (df[col] >= lower_bound) & (df[col] <= upper_bound)

df = df[mask]
```

```python
from scipy.stats.mstats import winsorize

kolom_belanja = ['belanja_buah', 'belanja_daging',
                 'belanja_ikan', 'belanja_kue', 'tahun_kelahiran']

for col in kolom_belanja:
    df[col] = winsorize(df[col], limits=[0.1, 0.1])
```

```python
#lihat statistik
df.describe().T
```

```python
df.plot(kind='box', subplots=True, layout=(-1,3), figsize=(12,8))
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 4, figsize=(20,20))
axes = axes.flatten()
idx = 0
for i in df.select_dtypes(exclude='object'):
    sns.histplot(df, x=i, ax=axes[idx], kde=True)
    axes[idx].set_title(f"Hist plot of {i}")
    idx +=1

plt.tight_layout()
plt.show()
```

```python
#melihat korelasi menggunakan method pearson. ubah valuenya jika ingin diubah ke kendall/spearman
corr = df.corr(method='pearson')

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='viridis', fmt=" .2f")
plt.title('Correlation Heatmap')
plt.show
```
